#### set methods
- Adding Elements to set
 - add(ele): it adds a single element to set if element not exist in set , otherwise  nothing
 - update(iterable): add all iterable elements to set
- Remove elements from set
 - remove(ele): removes the element if not there it raises key error
 - discard(ele): it removes the elements if it is present otherwise nothing
 - pop(): removes and returns the element from set
- set operations
 - union(another-sets):
 - intersection():
 - difference(): 
 - symmetric_difference(): union - intersection/ returns the non common elements from all sets 


```python
s1={1,2,3,4,5}
s2={"pavan","raju",3,4}
s3={2.5,3.5}
s1.add(6)
print(s1)
s1.remove(6)
print(s1)
s1.discard(6)
print(s1)
print(s1.union(s2,s3))
print(s1.intersection(s2,s3))
print(s1.difference(s2,s3))
print(s1.symmetric_difference(s2))
```

```python
s1={1,2,3,4,5}
s2={"pavan","raju",3,4}
s3={2.5,3.5}
print(s1|s2|s3) #union
print(s1&s2&s3) #intersection
print(s1-s2-s3)
print(s1^s2) #symmetric_difference
s1.update([4,5,6])
print(s1)
```
