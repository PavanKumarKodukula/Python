```python
#output formatting

name = "pavan"
age = 21
```

```python
# using +
print("my name is "+name+" and my age is "+str(age))
```

```python
#comma formatting
print("my name is",name,"and my age is",age)
```

```python
#by using access specifiers
print("my name is %s and my age is %d"%(name,age))
```

```python
#dot format
print("my name is {} and my age is {}".format(name,age))
```

```python
#f-format
marks = [99,100,98,97,96,95]
percentage=98.256
print(f"my name is {name} and my age is {age} and my percentage is {percentage}")
print(f"my name is {name} and my age is {age} and my percentage is {percentage:.2f}")
```
