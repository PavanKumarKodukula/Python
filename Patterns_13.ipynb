```python
#square shape * pattern
n=5
for i in range(n):
    for j in range(n):
        print("*",end=" ")
    print()
```

```python
n=5
for i in range(n):
    for j in range(n):
        if i==j or j==n-i-1:
            print("*",end=" ")
        else:
            print(" ",end=" ")    
    print()
```

```python
n=5
for i in range(n):
    for j in range(n):
        if j==0 or i==0 or j==n-1 or i==n-1:
            print("*",end=" ") 
        else:
            print(" ",end=" ")    
    print()
```
