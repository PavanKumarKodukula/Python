```python
def func(n):
    if n==0:
        return
    func(n-1)
    print(n)
func(3)   
```

```python
def fibonacci(n):
    if n==1 or n==0:
        return 1
    return fibonacci(n-1)+fibonacci(n-2)

fibonacci(4)
    
```

```python
# sum of n natural numbers
def sum_of_natural(n):
    if n==1:
        return 1
    return n+sum_of_natural(n-1)

print(sum_of_natural(5))
```

```python
def binary(n):
    if n==0:
        return "0"
    if n==1:
        return "1"
    return binary(n//2)+str(n%2)
binary(4)
```

```python
sum=lambda x,y:x+y
x,y=map(int,input().split())
print(sum(x,y))
```

```python
#find square of all list of integer using lambda and map()
arr=[1,2,3,4,5]
res=list(map(lambda x:x**2,arr))
print(res)
             
```

```python
arr=[1,2,3,4,5,6]
res=list(filter(lambda x:x%2==0,arr))
print(res)
```

```python
#single line if-else syntax
n=int(input())
res=lambda x:f"{x}-odd"if x%2 else f"{x}-even"
print(res(n))
```
