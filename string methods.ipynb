## String Methods
- #### Case Conversion Methods
   - Upper(): it converts the string into upper case
   - lower():converts the string into lower case
   - title(): starting letter of each word will be converted in capital
   - swapcase(): swap the lower to upper and vice versa
   - capitalize(): capitalize the 1st letter of the string

```python
String="The ocean is Blue7 and 8 sky is blue%"
print(String.lower())
print(String.upper())
print(String.title())
print(String.swapcase())
print(String.capitalize())
print("")
#Split 
print(String.split()) # split(): returns the list of substring and by default splits at white spaces
print(String.split("blue"))
print(String.split("z"))
print(String.split("%"))
```

```python
# White Space remove methods
# strip()
# lstrip():leading white space
# rstrip():lagging white space 
string="     pavan     "
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

```python
string="@@&*pavan*&^"
print(string.strip())
print(string.lstrip("@"))
print(string.rstrip("^"))
```

```python
# Indexing finding methods
# index(sub_string) : returns the 1st occurance if exists else it returns "value error"
# find(sub_string) :  if there is a error it returns -1 
string = "the ocean is blue and sky is blue"
print(string.index("e"))
print(string.index("is"))
print(string.index("is",11))#starts checking from 11th position
```

```python
string = "the ocean is blue and sky is blue"
print(string.find("e"))
print(string.find("is"))
print(string.find("is",11))
print(string.find("z"))
```

```python
# replace(old,new)
string = "the ocean is blue and sky is blue"
print(string.replace('blue','red'))
print(string.replace('blue','red',1))
print(string.replace('red','blue',1))
```

```python
# "sep".join(iterable): it joins all the iterable elements with seperator
lst = ['The', 'ocean', 'is', 'Blue', 'and','sky', 'is', 'blue']
print("".join(lst))
print(" ".join(lst))
print("_".join(lst))
```

### string checking methods
- islower():
- isupper():
- istitle():
- isalpha():
- isalnum():
- isdigit():
- startswith("substring"):
- endswith("substring"):

```python
print("hell0".islower())
print("HELL0".isupper())
print("Hello".istitle())
print("hello123".isalpha())
print("hello12".isalnum())
print("hello".startswith("He")) #case-sensitive
print("hello".endswith("Lo"))
```

```python
# count(sub_string) ; it returns count of non overlaping sub_string occurance
print("ABCCBCA".count('c'))
print("ABCBCB".count("BCB"))
```
