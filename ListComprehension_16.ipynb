```python
n=int(input())
lst=[i for i in range(1,n+1) if i%2==0]
print(lst)
```

```python
arr=[20,30,40,1,3,5]
lst=["odd" if i%2 else "even" for i in arr]
print(lst)
```

```python
arr=[20,30,40,1,3,5]
lst=[i+1 if i%2 else i+1 for i in arr]
print(lst)
```

```python
# find 2 dices combinations
print([(i,j) for i in range(1,7) for j in range(1,7)])
```
