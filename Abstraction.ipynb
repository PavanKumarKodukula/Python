```python
from abc import ABC,abstractmethod
class mobile(ABC):
    @abstractmethod
    def ram(self):
        pass
    @abstractmethod
    def rom(self):
        pass
    @abstractmethod
    def battery(self):
        pass
    def about_device(self):
        return "this is mobile"
class samsung(mobile):
    def __init__(self,model,samsung_ram,samsung_rom,samsung_battery):
        self.model=model
        self.samsung_ram=samsung_ram
        self.samsung_rom=samsung_rom
        self.samsung_battery=samsung_battery
    def ram(self):
        return f"the samsung {self.model} have {self.samsung_ram}gb ram"
    def rom(self):
        return f"the samsung {self.model} have {self.samsung_rom}gb ram"
    def battery(self):
        return f"the samsung {self.model} have {self.samsung_battery} mah"
    def camera(self,front,back):
        return f"the samsung {self.model} have {front}mp front camera and {back}mp back camera"

s23fe=samsung("s23fe",16,256,5000)
print(s23fe.about_device())
print(s23fe.ram())

    
```
