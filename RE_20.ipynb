```python
import re
string="The sky is blue and the ocean is blue"
res=re.match("[A-Z]",string)
print(res)
print(res.span())
print(res.group())
```

```python
# match(pattern, string)
string = "The sky is blue and ocean is blue"
res = re.match("[A-Z,a-z]+",string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched pattern value
```

```python
# match(pattern, string)
try:
    string = "The sky is blue and ocean is blue"
    res = re.match("z",string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern value
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
string="The sky is blue and ocean is blue"
res=re.search("[A-Za-z]{5}",string)
print(res)
print(res.span())
print(res.group())
```

```python
string="The sky is blue and ocean is blue"
res=re.findall("[A-Za-z]{3,}",string)
print(res)
```

```python
string="The sky is blue and oc12ean is b233lue"
res=re.findall("[A-Za-z]+",string)
print(res)
```

```python
#validate mobile number
try:
    mobile=input()
    res=re.fullmatch("^[6-9][0-9]{9}",mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is invalid")    
except Exception as e:
    print(f"{e}")
    
```

```python
try:
    pancard=input()
    res=re.fullmatch("^[A-Z]{5}[0-9]{4}[A-Z]{1}$",pancard)
    print(res)
    if res:
        print(f"{pancard} is valid")
    else:
        print(f"{pancard} is invalid")    
except Exception as e:
    print(f"{e}")
```

```python
import re
try:
    mail=input()
    res=re.fullmatch(r"^[a-z][a-z0-9.]+\@[a-z0-9.]+\.[a-z]{2,3}$",mail)
    print(res)
    if res:
        print(f"{mail} is valid")
    else:
        print(f"{mail} is invalid")    
except Exception as e:
    print(f"{e}")
```

```python
import re
try:
    date=input()
    res=re.fullmatch(r"^(0?[1-9]|[12][0-9]|3[01])[-/](0?[1-9]|1[0-2])[-/]([0-9]{4})$",date)
    print(res)
    if res:
        print(f"{date} is valid")
    else:
        print(f"{date} is invalid")    
except Exception as e:
    print(f"{e}")
```

```python
import re

try:
    password = input()

    res = re.fullmatch(
        r"^(?=.*[A-Z])(?=.*[a-z])(?=.*[0-9])(?=.*[^A-Za-z0-9]).{8,}$",
        password
    )

    print(res)

    if res:
        print(f"{password} is valid")
    else:
        print(f"{password} is invalid")

except Exception as e:
    print(f"{e}")
```
