```python
#dictionary
# if multiple keys names are similar , value will be the recently declared key's value
details={"name":"Pavan Kumar Kodukula","age":20,"course":["python full stack",60],"age":12,"Age":23}
print(details)
#access values by using keys
print(details["course"],details["age"])
# updating dictgionary values
details["age"]+=10
print(details["age"])
details["course"]="Python full stack developer"
print(details)
#item assignment 
details["location"]=["Hyderabad","remote"]
print(details)
```

### dictionary methods
- Get Values
  - get(key, default_val): it returns the value if key exists in dict otherwise default_val
- Update Values and add items to dictionary
  - update(new_dictionary): Dict updates with new-dictionary items
- Remove items from dict
  - pop(key): it returns value and remove item if key exists otherwise "key error"
  - popitem(): returns tuple of key value pair and removes the last item from dictionary
- Get keys , value and Items from dictionary
  - keys(): it returns list of dictionary keys
  - values():
  - items(): returns list of tuple key-value pairs 
  

```python
details={"name":"Pavan Kumar Kodukula","age":20,"course":["python full stack",60],"age":12,"Age":23}
#get method
#print(details["location"])
print(details.get("name"))
print(details.get("location"))
print(details.get("Salary","23.5lpa"))
print(details)
```

```python
details={"name":"Pavan Kumar Kodukula","age":20,"course":["python full stack",60],"age":12,"Age":23}
d={"marks":[99,99,100],"age":50}
#update method
details.update(d)
print(details)
```

```python
details={"name":"Pavan Kumar Kodukula","age":20,"course":["python full stack",60],"age":12,"Age":23}
#remove methods
print(details.pop("age"))
print(details)
print(details.popitem())
print(details)
print(details.keys())
print(details.values())
print(details.items())
# print(details.pop("percentage"))
```
