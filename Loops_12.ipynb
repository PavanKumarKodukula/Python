```python
# count vowels and consonants in a string
string = input()
vowels = consonants = 0
for alphabet in string:
    if alphabet in "aeiouAEIOU":
        vowels=vowels+1
    elif alphabet.isalpha():
        consonants=consonants+1   
print(f"vowels are {vowels} and consonents are {consonants} in {string}")
```

```python
# count even and odd numbers in given list of numbers
numbers=list(map(int,input().split()))
even=odd=0
for num in numbers:
    if num%2:
        odd=odd+1
    else:
        even=even+1
print(f"count of even numbers is {even} and count of odd numbers is {odd} in given {numbers} ")            
```

```python
# seperate the even and odd numbers in the given list
# lst = [1,2,3,4,5,6] o/p: even[2,4,6], odd[1,3,5]
numbers=list(map(int,input().split()))
even=[]
odd=[]
for num in numbers:
    if num%2:
        odd.append(num)
    else:
        even.append(num)
print(f"even numbers {even} ; odd numbers {odd} in  {numbers} ")     
```

```python
arr=list(map(int,input().split()))
target=int(input())
res=[]
n=len(arr)
i=0
while i<n:
    if arr[i] != target:
        res.append(arr[i])
    i=i+1
print(res)        
```

```python
arr=list(map(int,input().split()))
target=int(input())
n=len(arr)
i=0
while i<n:
    if arr[i] == target:
        arr.pop(i)
        n-=1
    else:
        i+=1
print(arr)        
```

```python
#return the target element position , if target doesnt exist return -1
arr=list(map(int,input().split()))
target=int(input())
n=len(arr)
for i in range(n):
    if arr[i] == target:
        print(i)
        break
    else:
        print(-1)
```

```python
num = int(input())
for i in range(2,num//2+1):
    if num%i==0:
        print(f"{num} is not prime")
        break
else:
    print(f"{num} is prime")    
```

```python
string=input()
rev=""
for character in string:
    rev=character+rev
print(rev)    
if rev==string:
    print(f"{string} is palindrome")
else:
    print(f"{string} is not palindrome")

```
