#### list methods
##### element adding methods
-   append(item) : add only single item at end of the list
-   extend(iterable): it adds all iterable elements at end of the list
-   insert(pos,item) : this method adds the item at the specified position

```python
lst=[1,2,3,4]
lst.append(10)
print(lst)
lst.append("code")
print(lst)
lst.append([10,20,30])
print(lst)
#print(lst[6])
lst.extend([100,200,30])
print(lst)
lst.insert(1,200)
print(lst)
```

#### Element remove methods
-  remove(val): it removes first occurence of value exists in lsit, otherwise returns 'value error'
-  pop(): remove and returns the last element of list
-  pop(pos): removes and returns the specified position element from list
-  clear(): removes all elements from list
-  del() : keyword that used to delete values

```python
lst=[10,20,30,20,40]
lst.remove(20)
#lst.remove(100) value error
print(lst)
val = lst.pop()
print(val,lst)
val2 = lst.pop(1)
print(val2, lst)
del lst[0]
print(lst)
lst.clear()
print(lst)
```

- index(val): returns the index of 1st occurence otherwise valuerror
