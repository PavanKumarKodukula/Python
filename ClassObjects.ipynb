```python
# updating instance attributr data from outside class
class Mobile:
    company="Samsung"
    def __init__(self,model,price,ram=8,rom=256):
        self.model=model
        self.ram=ram
        self.rom=rom
        self.price=price

samsungs23fe = Mobile("samsung23fe",30000)
samsungs24fe = Mobile("samsung24fe",30000,16)
print(samsungs23fe.ram)
samsungs23fe.ram=16
print(samsungs23fe.ram)
samsungs23fe.company="New Samsung"
print(samsungs23fe.company)
print(samsungs24fe.company)
```
