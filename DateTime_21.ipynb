```python
from datetime import datetime,timedelta
curr_timestamp=datetime.now()
print(curr_timestamp)
today=(curr_timestamp.date())
curr_time=(curr_timestamp.time())
print(today)
print(curr_time)
print(today.month)
print(today.year)
print(today.day)
print(curr_time.hour,curr_time.minute,curr_time.second)
print(today.strftime("%d-%m-%y"))
print(curr_time.strftime("%I:%M %p"))
```

```python
import time
from IPython.display import clear_output
minutes=int(input("Enter Stop Watch Minutes:"))
sec_counter = 0
min_counter = 0
hr_counter = 0

while min_counter!=minutes:
    time.sleep(1)
    sec_counter += 1
    
    if sec_counter == 60:
        sec_counter = 0
        min_counter += 1
        if min_counter == 60:
                min_counter = 0
                hr_counter += 1
        
    print(f"{hr_counter:02d}:{min_counter:02d}:{sec_counter:02d}")
    clear_output(wait=True)

```
